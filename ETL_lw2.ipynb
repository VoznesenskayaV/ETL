```python
import matplotlib.pyplot as plt

# сортировка по объёму продаж (по убыванию)
df_sorted = df.sort_values(by='total_sales', ascending=False)

# цвета по продажам
sales_sorted = df_sorted['total_sales']

def get_sales_color(value):
    if value == sales_sorted.max():
        return 'green'
    elif value == sales_sorted.min():
        return 'red'
    else:
        return 'gold'

colors = df_sorted['total_sales'].apply(get_sales_color)

plt.figure(figsize=(8, 5))
bars = plt.bar(df_sorted['category'], df_sorted['total_sales'], color=colors)

plt.title('Объём продаж по категориям (2016)')
plt.xlabel('Категория')
plt.ylabel('Продажи')

# подписи
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height,
        f'{height:,.0f}',
        ha='center',
        va='bottom'
    )

plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# функция для цвета
def get_color(margin):
    if margin >= 0.13:
        return 'green'
    elif margin >= 0.08:
        return 'gold'
    else:
        return 'red'

colors = df['profit_margin'].apply(get_color)

plt.figure(figsize=(8, 5))
bars = plt.bar(df['category'], df['profit_margin'], color=colors)

plt.title('Маржинальность по категориям (2016)')
plt.xlabel('Категория')
plt.ylabel('Profit margin')

# подписи над столбцами
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height,
        f'{height:.2f}',
        ha='center',
        va='bottom'
    )

plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```
